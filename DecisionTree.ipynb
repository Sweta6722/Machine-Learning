### Q1. What is a Decision Tree? Explain its structure (Root, Nodes, Leaves) with a real-life example.

Ans: A Decision Tree is a supervised machine learning algorithm used for classification and regression tasks. It makes decisions by splitting data into smaller groups based on specific conditions, forming a tree-like structure.

The model starts with a question and continues asking further questions until it reaches a final prediction or decision.

**Structure of a Decision Tree**

A Decision Tree consists of three main parts:

**1. Root Node**

- The topmost node of the tree.
- Represents the entire dataset.
- It is the first decision point from which the tree starts.

Example:
“Is the weather sunny?”

**2. Decision Nodes (Internal Nodes)**

- Nodes that represent conditions or questions.
- They split the data into different branches based on the answer.

Example:
“Is humidity high?”
“Is temperature above 30°C?”

**3. Leaf Nodes (Terminal Nodes)**

- The final nodes of the tree.
- They provide the prediction or outcome.
- No further splitting occurs.

Example:
“Play Cricket” or “Don’t Play Cricket”

**Real-Life Example: Deciding Whether to Play Cricket**

**Explanation**

- Root Node: Weather?
- If weather is Rainy, the decision is Don’t Play Cricket.
- If weather is Sunny, check Humidity.
- If humidity is High, Don’t Play Cricket.
- If humidity is Low, Play Cricket.

**Advantages of Decision Trees**

- Easy to understand and visualize.
- Requires little data preparation.
- Works for both classification and regression.
- Handles numerical and categorical data.

**Disadvantages of Decision Trees**

- Can overfit the training data.
- Small changes in data may create a different tree.
- Complex trees can become difficult to interpret.

**Conclusion**

A Decision Tree is a simple and interpretable machine learning model that makes predictions through a series of decision rules. It consists of a Root Node, Decision Nodes, and Leaf Nodes, which together form a tree-like structure for making decisions.

### Q2. Differentiate between Gini impurity and Entrophy. Which one is used by default in Scikit-learn and why?

Ans: **Differentiate between Gini impurity and Entrophy**

Both Gini impurity and Entrophy are criteria used in Decision Trees to measure how well a split separates the data into classes.

**Gini Impurity**

* Definition - Measure the probability of incorrectly classifying a randomly chosen sample.
* Range (Binary Class) - 0 to 0.5
* Pure Node Value - 0
* Computation Speed - Faster
* Mathematical Operation - Uses squaring
* Split Selection - Tends to isolate the largest class

**Entropy**

* Definition - Measure the amount of uncertainty or disorder in the dataset.
* Range (Binary Class) - 0 to 1
* Pure Node Value - 0
* Computation Speed - Slightly slower
* Mathematical Operation - Uses logarithms
* Split Selection - Produces more balanced trees

**Which one is used by default in Scikit-learn?**

In the Scikit-learn DecisionTreeClassifier, the default criterion is:

* DecisionTreeClassifier()

which is equivalent to:

* DecisionTreeClassifier(criterion='gini')

**Why is Gini the Default?**

- Computationally faster because it does not require logarithmic calculations.
- Produces results very similar to Entropy in most practical problems.
- More efficient on large datasets, making model training faster.
- Historically, it has been the standard choice in many CART (Classification and Regression Tree) implementations.


### Q3. What is Overfitting in Decision Trees? How can we detect it using training and testing accuracy?

Ans: Overfitting occurs when a Decision Tree learns not only the important patterns in the training data but also the noise and random fluctuations.

As a result:

- The model performs very well on training data.
- The model performs poorly on new (unseen) test data.
- The tree becomes overly complex with many branches and leaves.

**Example**

Suppose a Decision Tree is built to predict whether a customer will buy a product.

- A simple tree learns general patterns and works well on new customers.
- An overfitted tree memorizes specific details of the training customers, making it less effective on new customers.

**How to Detect Overfitting Using Training and Testing Accuracy?**

Compare the model’s accuracy on the training dataset and the testing dataset.

**Training Accuracy**

Situation:

* Good Fit - High
* Underfitting - Low
* Overfitting - Very High (often 100%)

**Testing Accuracy**

Situation:

* Good Fit - High
* Underfitting - Low
* Overfitting - Much Lower

**Interpretation**

Situation:

* Good Fit - Model generalizes well
* Underfitting - Model is too simple
* Overfitting - Model memorized trainind data

**Example**

* Training Accuracy = 99%
* Testing Accuracy = 72%

The large gap (27%) indicates overfitting.

**Another Example**

* Training Accuracy = 99%
* Testing Accuracy = 88%

 The small gap (2%) indicates the model is generalizing well.

### Q4. Explain Pruning in Decision Trees. What is the difference between Pre-pruning and Post-pruning?

Ans: **Pruning in Decision Trees**

Pruning is a technique used to reduce the size of a Decision Tree by removing unnecessary branches that do not significantly improve prediction accuracy.

The main purpose of pruning is to prevent overfitting and improve the model’s ability to generalize to new data.

**Why is Pruning Needed?**

A fully grown Decision Tree may:

- Fit the training data very closely.
- Capture noise and outliers.
- Perform poorly on unseen test data.

Pruning simplifies the tree by removing less important branches, resulting in:

- Better generalization.
- Reduced complexity.
- Faster predictions.

**Types of Pruning**

1. **Pre-pruning (Early Stopping)**

In Pre-pruning, the growth of the tree is stopped before it becomes too complex.

The algorithm decides not to split a node if certain conditions are met.

**Common Stopping Criteria**

- Maximum tree depth reached (max_depth)
- Minimum samples required to split (min_samples_split)
- Minimum samples in a leaf node (min_samples_leaf)
- Minimum information gain threshold

**Example**

Suppose a node contains only a few records. Instead of creating more branches, the algorithm stops splitting that node.

**Advantages**

- Faster training.
- Reduces overfitting.
- Produces smaller trees.

**Disadvantages**

- May stop too early.
- Can lead to underfitting.

2. **Post-pruning (Backward Pruning)**

In Post-pruning, the tree is first grown completely and then unnecessary branches are removed.

The algorithm evaluates whether removing a branch improves performance on validation data.

**Process**

- Build the full tree.
- Evaluate each subtree.
- Remove branches that do not improve accuracy.
- Keep the simpler tree.

**Example**

If a branch classifies only a few training samples and does not improve validation accuracy, it is removed.

**Advantages**

- Usually gives better accuracy.
- Less risk of underfitting.
- Produces an optimal-sized tree.

**Disadvantages**

- More computationally expensive.
- Requires building the full tree first.

**Difference Between Pre-pruning and Post-pruning**

**Pre-pruning**

* When applied? - During tree construction
* Approach - Stops tree growth early
* Computational Cost - Lower
* Risk - Underfitting
* Training Time - Faster
* Accuracy - May be lower if stopped too early
* Tree Size - Smaller from the beginning

**Post-pruning**

* When applied? - After tree construction
* Approach - Remove branches from a fully grown tree
* Computational Cost - Higher
* Risk - Less risk of underfitting
* Training Time - Slower
* Accuracy - Often higher
* Tree Size - Reduced after full growth


### Q5. What is Feature Importance? How can it help businesses in decision-making?

Ans: Feature Importance is a technique used in Machine Learning to measure how much each input variable (feature) contributes to a model’s predictions.

Feature importance can show that:

* Credit Score – 45%
* Income – 30%
* Employment Status – 15%
* Age – 10%

This means the model relies most heavily on Credit Score when making predictions.

**Why is Feature Importance Useful?**

1. **Better Business Understanding**

It helps businesses identify the key factors driving outcomes.

**Example:**

A telecom company predicting customer churn may find that:

* Contract type
* Monthly charges
* Customer service complaints

 These are the most important factors influencing customer departures.

2. **Improved Decision-Making**

Managers can focus on the factors that matter most.

**Example:**

A bank learns that low credit scores are the strongest indicator of loan default. It can tighten credit checks and reduce risk.

3. **Cost Reduction**

Businesses can remove unimportant features, reducing data collection and storage costs.

**Example:**

If 20 customer attributes are collected but only 8 significantly affect predictions, the company may stop collecting the less useful ones.

4. **Better Model Performance**

Removing irrelevant features can:

* Reduce overfitting
* Speed up training
* Improve model interpretability

5. **Explainability and Trust**

Feature importance helps explain why a model made a particular prediction, making it easier for stakeholders to trust the results.

**Feature Importance in Decision Trees**

Decision Trees naturally calculate feature importance based on how much each feature reduces impurity (using measures such as Gini Impurity or Entropy).

Features that create the most useful splits receive higher importance scores.

```python
### Q8. Data Preprocessing
# Convert categorical variables into numerical
# Mention which encoding method you used

# Interpret:
# Why is encoding necessary for Decision Trees?

# Convert Categorical Variables

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.select_dtypes(include='object'):
    df[col] = le.fit_transform(df[col])
```

## Interpretation

Encoding method used: Label Encoding

Label Encoding converts each category into a unique numerical value.

Machine learning algorithms require numerical input. Therefore, categorical variables were converted into numerical form using encoding techniques. This transformation allows the Decision Tree model to process and learn patterns from all variables effectively.

**Why is encoding necessary for Decision Trees?**

Decision Trees cannot directly interpret text values such as “yes”, “no”, or job categories. Encoding converts these categories into numerical values that the algorithm can use for splitting and classification.

```python
### Q9. Feature Selection & Splitting
# Separate:
# Feature (X)
# Target (y)
# Perform train-test split (80-20)

# Interpret:
# Why is train-test split important?

# Separate X and y

X = df.drop('deposit', axis=1)
y = df['deposit']
```

```python
# Train-Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)
```

## Interpretation

The dataset was divided into features (X) and target variable (y). An 80-20 train-test split was performed, where 80% of the data was used for training and 20% for testing. This helps evaluate the model on unseen data and provides a realistic measure of performance.

**Why is train-test split important?**

Train-test splitting helps assess how well the model generalizes to new data. Without a separate test set, it would be difficult to determine whether the model is truly learning patterns or simply memorizing the training data.

```python
### Q10. Model Building
# Train a Decision Tree Classifier
# Use:
# criterion='gini'
# max_depth=5

# Interpret:
# What does max_depth control?

# Train Decision Tree

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)
```

## Interpretation

A Decision Tree Classifier was trained using the Gini criterion with a maximum depth of 5. The model learned patterns from customer data to predict whether a customer would subscribe to a term deposit.

**What does max_depth control?**

The max_depth parameter controls the maximum number of levels in the tree. Limiting the depth helps prevent overfitting and improves the model’s ability to generalize to new data.

```python
### Q11. Model Evaluation
# Perform:
# Accuracy score
# Confusion Matrix
# Classification Report

# Interpret (MUST):
# Is the model performing well?
# Explain Precision & Recall
# Any class imbalance issue?

# Predictions

y_pred = model.predict(X_test)
```

```python
# Accuracy

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(accuracy)
```

```python
# Confusion Matrix

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
# Classification Report

from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

## Interpretation

The model was evaluated using Accuracy Score, Confusion Matrix, and Classification Report.

**Accuracy:**

- Accuracy measures the overall correctness of predictions.

**Precision:**

- Precision indicates how many predicted positive cases were actually positive.

**Recall:**

- Recall indicates how many actual positive cases were correctly identified.

**F1 Score:**

- F1-Score provides a balance between Precision and Recall.

A higher value for these metrics indicates better model performance.

**Is there any class imbalance issue?**

If one class (Yes or No) has significantly more observations than the other, the dataset may be imbalanced. In such cases, Accuracy alone may not be sufficient, and Precision, Recall, and F1-score should also be considered.

```python
### Q12. Overfitting Check:
# Calculate:
# Training accuracy
# Testing accuracy

# Interpret:
# Is the model overfitting or underfitting?

# Training Accuracy

train_acc = model.score(X_train, y_train)
print(train_acc)
```

```python
# Testing Accuracy

test_acc = model.score(X_test, y_test)
print(test_acc)
```

## Interpretation:

Difference is small.

Therefore, Model is not overfitting.

Training accuracy and testing accuracy were compared to check model generalization.

- If training accuracy is much higher than testing accuracy, the model is overfitting.
- If both accuracies are low, the model is underfitting.
- If both accuracies are high and similar, the model is performing well.

This comparison helps determine whether the model has learned meaningful patterns or memorized the training data.
